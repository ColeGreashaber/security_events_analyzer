```python
from security_events import SecurityEvent, LoginEvent, MalwareAlert, NetworkEvent


# Events Created 
event1 = LoginEvent("192.168.1.25", "High", "administrator")
event2 = LoginEvent("10.0.0.15", "Critical", "root")

event3 = MalwareAlert("172.16.0.20", "Medium", "suspicious.exe")
event4 = MalwareAlert("10.0.0.77", "Critical", "payload.exe")

event5 = NetworkEvent("192.168.1.100", "High", "8.8.8.8", 4444)
event6 = NetworkEvent("10.10.10.25", "Critical", "185.10.10.50", 4444)
event7 = NetworkEvent("192.168.1.5", "Low", "8.8.8.8", 443)


# Events Stored
critical_count = 0
high_count = 0
medium_count = 0
low_count = 0  

# Counters
events = [ event1, event2, event3, event4, event5, event6, event7]
for event in events:
    if event.is_critical() == True:
        critical_count = critical_count +1
        print("Critical Event Detected!\n")

    elif event.severity == "High":
        high_count = high_count +1

    elif event.severity == "Medium":
        medium_count = medium_count +1
    elif event.severity == "Low":
        low_count = low_count +1
        event.show_event()
        print()

#Finding SUspicious Ports
for event in events:
    if isinstance(event, NetworkEvent):
        if event.is_suspicious_port():
            print("SUSPICIOUS PORT DETECTED")
            print(f"Source_ip: {event.source_ip}")
            print(f"Destination_ip: {event.destination_ip}")
            print(f"Port: {event.port}")
            print()



print("SECURITY EVENT SUMMARY")
print()

print(f"Total Events {len(events)}")
print(f"Critical Events: {critical_count}")
print(f"High Events: {high_count}")
print(f"Medium Events: {medium_count}")
print(f"Low Events: {low_count}")
```
